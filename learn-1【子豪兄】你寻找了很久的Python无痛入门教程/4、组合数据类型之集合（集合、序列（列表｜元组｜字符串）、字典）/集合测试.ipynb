集合的运算

```python
s = {'上海','交大','同济','复旦','浙大','清华','北大'}
t = {'武大','华工','理工大','华农'}
```

```python
s & t
```

```python
s-t
```

```python
s<t
```

```python
s>t
```

```python
# 添加
s.add('山东大学')
# 删除
s.remove('交大')
```
