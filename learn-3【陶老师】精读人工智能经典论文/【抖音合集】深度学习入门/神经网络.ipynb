```python
import torch
import torch.nn.functional as F
```

```python
## 定义样本数值
x1=torch.tensor(1)
x2 = torch.tensor(2)
Y= torch.tensor(2)
# 参数初始化
w11=torch.tensor(1.,requires_grad=True)
w12 =torch.tensor(1.,requires_grad=True)
w21=torch.tensor(1.,requires_grad=True)
w22=torch.tensor(-1.,requires_grad=True)
w1=torch.tensor(1.,requires_grad=True)
w2=torch.tensor(-1.,requires_grad=True)
b11,b12,b21=torch.tensor([1.,1.,1.],requires_grad=True)
lr=0.01
```

```python
# 前向传播
z1 = w11*x1 + w12*x2 + b11
a1 = F.relu(z1)
z2 = w21*x1 +w22*x2 + b21
a2 =F.relu(z2)
y= w1*a1 + w2*a2 + b21
```

```python
# 求损失
loss = (y-Y).pow(2) / 2
```

```python
# 求参数梯度
w11_grad = (y-Y)*w1*1*x1
w11_grad
```

```python
# w11参数更新参数
#w11 = w11 - lr*w11_grad
#w11
```

```python
# 所有参数批量更新
loss.backward()
```

```python
w11.grad
```

```python
with torch.no_grad():
    w11 -= lr*w11.grad
```
