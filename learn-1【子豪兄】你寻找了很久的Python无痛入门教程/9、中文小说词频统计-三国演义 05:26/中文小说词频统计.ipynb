```python
import jieba
```

```python
f = open('三国演义.txt', 'r', encoding='utf-8')
```

```python
txt = f.read()
```

```python
words=jieba.lcut(txt)
```

```python
words
```

```python
len(words)
```

```python
len(set(words))
```

```python
# 统计
counts = {}
for word in words:
    if len(word) == 1:  # 排除单个字符的分词结果
        continue
    else:
        counts[word] = counts.get(word,0) + 1
```

```python
counts
```

```python
# 转程list列表
items = list(counts.items())
# 进行排序
items.sort(key = lambda x:x[1], reverse = True)
```

```python
items
```

```python
for i in range(10):
    word, count = items[i]
    print("{0:<10}{1:>5}".format(word, count))
```
