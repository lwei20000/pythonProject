```python
import jieba
```

```python
#精确模式
jieba.lcut('中国科学院大学的学生')
```

```python
# 全模式
jieba.lcut('中国科学院大学的学生', cut_all=True)
```

```python
# 需要安装百度的paddlepaddle：pip install paddlepaddle
jieba.enable_paddle()
jieba.cut('中国科学院大学的学生', use_paddle=True)
```

```python
jieba.add_word('科学院大学')
jieba.lcut('中国科学院大学的学生', cut_all=True)
```
