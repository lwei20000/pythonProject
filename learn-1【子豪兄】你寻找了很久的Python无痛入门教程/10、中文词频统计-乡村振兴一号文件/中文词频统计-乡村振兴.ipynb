```python
import jieba
```

```python
# 导入外部文件
f=open("关于实施乡村振兴战略的意见.txt", "r", encoding='utf-8')
```

```python
# 读取文件并
txt=f.read()
```

```python
# 通过jieba分词
words = jieba.lcut(txt)
```

```python
words
```

```python
len(words)
```

```python
len(set(words))
```

构建词频映射字典


```python
counts={}
for word in words:
    # 这里是限制词语长度
    if len(word) == 4:
        counts[word]=counts.get(word,0)+1
```

```python
counts
```

```python
# 转换成列表
items=list(counts.items())
```

```python
items
```

```python
#词频按降序排序
items.sort(key=lambda x:x[1],reverse=True)
```

```python
items
```

```python
#打印输出排名前15的词语
for i in range(15):
    word,count=items[i]
    print("{0:<10}{1:>5}".format(word,count))
```
